```python
#list de dict
#dict contém "id" para cada "name"

users = [
  {"id": 0, "name": "Hero"},
  {"id": 1, "name": "Dunn"},
  {"id": 2, "name": "Sue"},
  {"id": 3, "name": "Chi"},
  {"id": 4, "name": "Thor"},
  {"id": 5, "name": "Clive"},
  {"id": 6, "name": "Hicks"},
  {"id": 7, "name": "Devin"},
  {"id": 8, "name": "Kate"},
  {"id": 9, "name": "Klein"}
  ]
```

```python
#exibindo toda a lista users

print(users)
```

```python
#exibindo o elemento da posição 1 da lista users

print(users[1])
```

```python
#exibindo o valor específico de uma chave da lista de dicionários

print("Nome:", users[1]['name'])
print("ID:", users[1]['id'])
```

```python
#lista de tuplas que indicam as relações de amizade pelos "id"

friendships = [(0,1),(0,2),(1,3),(1,3),(2,3),(3,4),
               (4,5),(5,6),(6,8),(7,8),(8,9)]
```

```python
#criando uma lista de amigos para cada usuário

for user in users: #para cada usuário da lista
  user["friends"] = [] #criamos uma lista vazia
```

```python
#povoando a lista com com os dados de "friendships"

for i,j in friendships:
  #isso funciona porque users[i] é o usuário cuja id é i
  users[i]["friends"].append(users[j]) #adiciona i como amigo de j
  users[j]["friends"].append(users[i]) #adiciona j como amigo de i
```

```python
#encontrando o tatal de conexões

def number_of_friends(user):
  """quantos amigos o usuário tem?"""
  return(len(user["friends"])) #tamanho da lista friend_ids

total_connections = sum(number_of_friends(user) for user in users)

print(total_connections)
```

```python
#número médio de conexões

from __future__ import division
num_users = len(users)
avg_connections = total_connections / num_users

print(avg_connections)
```
